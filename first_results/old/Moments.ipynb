```python
def clean_name(x):
    return x.replace("(","").replace(")","").replace("=","").replace("+","p").replace("/","")
```

```python
basedir = '/orange/adamginsburg/brick_alma_linesurvey/abulatek/first_results/figures/moments_FINAL'
# basedir = '/orange/adamginsburg/brick_alma_linesurvey/abulatek/first_results/figures/moments_alyssa_contd'
# otherdir1 = '/orange/adamginsburg/brick_alma_linesurvey/abulatek/first_results/figures/moments_alyssa'
# otherdir2 = '/orange/adamginsburg/brick_alma_linesurvey/abulatek/first_results/figures/moments'

vmin = -10*u.km/u.s
vmax = 70*u.km/u.s
for fn in fns:
    cube = SpectralCube.read(fn, use_dask=True).with_spectral_unit(u.GHz)

    slc  = cube[:1].subcube_slices_from_mask(cube[:1].mask)
    cube = cube[:,slc[1],slc[2]]
    cube.beam_threshold = 10000

    for row in tbl:
        restf = u.Quantity(row['Frequency (GHz)'], u.GHz)
        if cube.spectral_axis.min() < restf and cube.spectral_axis.max() > restf:
            outfn = f"{clean_name(row['Species'])}_{row['Frequency (GHz)']:0.4f}_{row['Dataset']}"
            filename_to_test = outfn.split('data')[0]
            if not glob.glob(f'{basedir}/{filename_to_test}*.max.png'):
            # if not ((glob.glob(f'{basedir}/{filename_to_test}*.max.png')) or (glob.glob(f'{otherdir1}/{filename_to_test}*.max.png')) or (glob.glob(f'{otherdir2}/{filename_to_test}*.max.png'))):
#             if not ((os.path.exists(f'{basedir}/{outfn}.max.png')) or (os.path.exists(f'{otherdir1}/{outfn}.max.png')) or (os.path.exists(f'{otherdir2}/{outfn}.max.png'))):
# The above doesn't work because I changed the convention for filenames between creating these files. data31 (old) vs. data_31 (new)
            #if True:
                print(outfn)
                scube = cube.with_spectral_unit(u.km/u.s, rest_value=restf, velocity_convention='radio').spectral_slab(vmin, vmax)
                mx = scube.max(axis=0)
                mx.quicklook(f'{basedir}/{outfn}.max.png')
                m0 = scube.moment0(axis=0)
                m0.quicklook(f'{basedir}/{outfn}.mom0.png')
```
